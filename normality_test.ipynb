```python
import pandas as pd
import os

if os.path.exists('data.csv'):
    df = pd.read_csv('data.csv')
else:
    df = pd.DataFrame({'Measurement': [1.2, 2.3, 1.8, 2.5, 3.1]})
    print("data.csv not found; using sample data instead.")
```

```python
# Create a figure with two subplots
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import pandas as pd

# df is already defined in a previous cell

plt.figure(figsize=(12, 5))

# Plot 1: Histogram
plt.subplot(1, 2, 1)
sns.histplot(df['Measurement'], kde=True, color='skyblue', stat='density')
plt.xlabel('Measurement')
plt.ylabel('Frequency')
plt.title('Histogram with KDE Curve')

# Plot 2: Q-Q Plot
plt.subplot(1, 2, 2)
stats.probplot(df['Measurement'], dist="norm", plot=plt)
plt.title('Q-Q Plot')

plt.tight_layout()
plt.show()
```
